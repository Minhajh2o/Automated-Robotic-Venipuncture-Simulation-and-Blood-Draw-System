```python
import yaml

dataset_path = "/kaggle/input/vein-yolov11"  # match your uploaded dataset's folder name

data_yaml = {
    "train": f"{dataset_path}/train/images",
    "val": f"{dataset_path}/valid/images",
    "nc": 1,
    "names": ["Vein"]
}

with open("/kaggle/working/data.yaml", "w") as f:
    yaml.dump(data_yaml, f)
```

```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")  # nano model — fast, good fit for 83 images

results = model.train(
    data="/kaggle/input/datasets/minhaj099/vein-yolov11/Vein.v1i.yolov11/data.yaml",
    epochs=300,
    imgsz=640,
    batch=16,
    patience=50,       # early stop if val loss plateaus
    project="/kaggle/working/runs",
    name="vein_detect"
)
```

```python
from ultralytics import YOLO

# load your trained weights, not the base yolo11n.pt
model = YOLO("/kaggle/working/weights/yolo26n.pt")

results = model.predict(
    source="/kaggle/input/datasets/minhaj099/testv1/person_001_db1_L2_png.rf.36956b13fa9afb6822fb94a5532e27c6.jpg",
    conf=0.25,       # confidence threshold — lower it if it's missing real veins, raise it if too many false positives
    save=True        # saves annotated output image
)
```

```python
import matplotlib.pyplot as plt

for r in results:
    im_array = r.plot()  # returns numpy array with boxes drawn
    plt.figure(figsize=(10, 10))
    plt.imshow(im_array[..., ::-1])  # BGR -> RGB
    plt.axis("off")
    plt.show()
```

```python
for r in results:
    for box in r.boxes:
        xyxy = box.xyxy[0].tolist()     # [x1, y1, x2, y2]
        conf = box.conf[0].item()
        print(f"Vein detected at {xyxy} with confidence {conf:.2f}")
```
